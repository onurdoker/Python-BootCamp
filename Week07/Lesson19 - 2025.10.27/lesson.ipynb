# 1. What is NumPy? Why do we use it?

* NumPy (Numerical Python) is a fundamental library in Python for scientific computing.

* It provides high-performance, multidimensional arrays and matrices along with tools to work with them:

* ndarray: Multidimensional, homogeneous, and high-performance array objects.

## Why not use regular lists instead?
1. Speed: NumPy is implemented in C under the hood, providing faster access to elements and enabling vectorized operations without loops.
2. Functions: It comes with built-in functions for linear algebra, statistics, Fourier transforms, etc., making complex computations more efficient.
3. Broadcasting: NumPy allows easy element-wise operations between arrays of different shapes through a feature called broadcasting.

# Introduction to NumPy: Why is it needed?

| Özellik    | Python Array                                                                        | NumPy Array (ndarray)                                                      |
|------------|-------------------------------------------------------------------------------------|----------------------------------------------------------------------------|
| Data Type  | Can hold mixed types (like int, string, float) at the same time.                    | Contains only a single type of data (all integers, all floats, etc.).      |
| Speed      | Slow. Each element is stored separately; looping is required to perform operations. | Very fast. All elements are stored side by side in a single memory block.       |
| İşlem Tipi | Cannot perform element-wise mathematical operations.                         | Can perform vectorized operations (executes thousands of operations simultaneously with a single command). |



## Analogies:

- Python List: A shopping cart containing items of various shapes and sizes, where you have to individually inspect and process each item
- NumPy Array: A well-organized package containing only identical items (e.g., all apples) with the same weight. You can process th entire package at once.

# Broadcasting

A mechanism that enables operations between arrays od different shapes.


When NumPy performs operations between arrays of different shapes, it automatically broadcasts the smaller (or one-dimensional) array, effectively expanding it to match the larger shape as if it were replicated

_Core Rule: The smaller array is broadcast (conceptually repeated) to match the shape of the larger array. Array dimensions are compared from right to left. Two dimensions are compatible if they are equal or if one of them is 1._


```python
a = np.array([10, 20, 30])  # [10, 20, 30]
b = 5  # scalar value ->      [5, 5, 5]
print(a + b)
```

```python
matrix_ = np.array([[1, 2, 3], [4, 5, 6]])
print("matrix: ", matrix_)
vector_ = np.array([10, 20, 30])
print("Vector: ", vector_)  # -> [10, 20, 30]  -> [10, 20, 30], [10, 20, 30]
result = matrix_ + vector_
print("Result: ", result)
```

## Broadcasting Rules (Compatibility Conditions)

For two arrays to be broadcastable, their shapes are compared from right to left. For each corresponding dimension, one of the following must hold:
1. The dimensions are equal (e.g., 3 and 3)
2. One of the dimensions is 1 (e.g., 5 and 1, or 1 and 7)

Otherwise, NumPy raises a ValueError: operands could not be broadcast together....

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Original Matrix:\n", matrix)
print("Matrix + 10:\n", matrix + 10)

print("---------")
a = np.array([[1, 2, 3], [4, 5, 6]])  # shape: (2, 3)
b = np.array([10, 20])  # shape: (2,)

try:
    print(
        a + b
    )  # ValueError: operands could not be broadcast together with shapes (2,3) (2,)
except ValueError as e:
    print("HATA:", e)
```

```python
# Storing temperature, humidity, and wind speed measurements from a weather station as a NumPy array.

# One-Dimensional Array: A single series of sequential measurements from a device (e.g., temperatures)
temperatures = np.array([22.5, 23.1, 22.9, 23.5, 24.0])
print(f"Daily Temperatures: {temperatures}°C")

# Two-Dimensional Array: Simultaneous measurements from different sensors (Matrix) [Temperature, Humidity, Wind Speed]
weather_data = np.array([[22.5, 55.0, 15.0], [23.1, 52.5, 18.0], [22.9, 53.0, 16.5]])
print(f"\nMulti-sensor weather data:\n {weather_data}")
```

```python
# Converting Celsius to Fahrenheit using NumPy arrays.
celsius_temperature = np.array([10.0, 15.0, 20.0])
fahrenheit_temperature = (celsius_temperature * 9 / 5) + 32
print(f"Fahrenheit Temperatures: {fahrenheit_temperature}")
```

```python
# Converting Celsius to Fahrenheit using a loop (Python list)
celsius_temperature = [10, 15.0, 20.0]
fahrenheit_temperature = []
for temperatures in celsius_temperature:
    fahrenheit_temperature.append(temperatures * 9 / 5 + 32)

print(f"Fahrenheit Temperatures: {fahrenheit_temperature}")
```
