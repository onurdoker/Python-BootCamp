# **Pandas DataFrame**

DataFrame is the most important data structure of the pandas library. It is a 2-dimensional, labeled data structure like an Excel table.

DataFrames are structures formed by the combination of series. They are often likened to SQL tables with columns and rows of different types. DataFrames allow us to process data more easily. DataFrames are often likened to tables in databases. They consist of Columns and indexes (rows). Data analysis can be performed on these values.

# Creating DataFrame with Dictionary (most common method)

```python
# Most common method
data = {
    "Name": ["Jack", "Jim", "John", "Jill", "Jenny"],
    "Age": [25, 30, 35, 28, 32],
    "City": ["New York", "Boston", "Los Angles", "Miami", "Chicago"],
    "Salary": [50000, 60000, 70000, 55000, 65000],
}

df = pd.DataFrame(data)
print("DataFrame created with dictionary:")
print(df)
print(f"Size: {df.shape}")  # (5, 4)
print(f"Shape: {df.shape[0]} rows, {df.shape[1]} columns")
```

# Accessing General DataFrame Properties

```python
print("\n" + "=" * 50)
print("BASIC DATAFRAME PROPERTIES")
print("=" * 50)

data = {
    "Name": ["Jack", "Jim", "John", "Jill", "Jenny"],
    "Age": [25, 30, 35, 28, 32],
    "City": ["New York", "Boston", "Los Angles", "Miami", "Chicago"],
    "Salary": [50000, 60000, 70000, 55000, 65000],
}

df = pd.DataFrame(data)

print("DataFrame info:")
print(df.info())

print(f"\nSize: {df.shape}")
print(f"Row count: {len(df)}")
print(f"Column count: {len(df.columns)}")
print(f"Column names: {list(df.columns)}")
print(f"Index info: {df.index}")
print(f"Data types:\n{df.dtypes}")

print("\nFirst 3 rows:")
print(df.head(3))

print("\nLast 2 rows:")
print(df.tail(2))
```

# Column Operations

```python
print("\n" + "=" * 50)
print("COLUMN OPERATIONS")
print("=" * 50)

# Selecting column
print("Name column:")
print(df["Name"])
print(type(df["Name"]))  # pandas Series

# Selecting multiple columns
print("\nName and Age columns:")
print(df[["Name", "Age"]])

# Adding new column
df["Experience"] = [3, 5, 8, 4, 6]
df["Annual Salary"] = df["Salary"] * 12
df["Age Group"] = np.where(df["Age"] < 30, "Young", "Adult")

print("\nNew columns added:")
print(df)

# Deleting column
df_drop = df.drop("Age Group", axis=1)  # axis=1 means column
print("\nAge Group column dropped:")
print(df_drop.columns)
```

# Row Operations

```python
print("\n" + "=" * 50)
print("ROW OPERATIONS")
print("=" * 50)

# Selecting row - .loc and .iloc
print("First row (.iloc):")
print(df.iloc[0])

print("\nFirst 3 rows:")
print(df.iloc[:3])

print("\nRow at specific index (.loc):")
print(df.loc[2])  # row at index 2

# Selecting rows by condition
print("\nThose older than 30:")
print(df[df["Age"] > 30])

print("\nThose living in Istanbul:")
print(df[df["City"] == "Istanbul"])

print("\nThose living in Boston:")
print(df[df["City"] == "Boston"])

# Multiple condition
print("\nThose older than 30 AND salary more than 60000:")
print(df[(df["Age"] > 30) & (df["Salary"] > 60000)])
```

# Index Operations

```python
print("\n" + "=" * 50)
print("INDEX OPERATIONS")
print("=" * 50)

# Setting custom index
df_custom = df.copy()
df_custom.index = ["A", "B", "C", "D", "E"]
print("DataFrame with custom index:")
print(df_custom)

# Resetting index
df_reset = df_custom.reset_index()
print("\nIndex reset:")
print(df_reset)

# Making index a column
df_custom = df_custom.reset_index().rename(columns={"index": "ID"})
print("\nIndex became column:")
print(df_custom)

# Making column an index
df_indexed = df.set_index("Name")
print("\nName column made index:")
print(df_indexed)
```

# Data Filtering and Querying

```python
print("\n" + "=" * 50)
print("DATA FILTERING AND QUERYING")
print("=" * 50)

# .query() method
print("Querying with .query():")
print(df.query("Age > 30 and Salary > 60000"))

# Filtering with .isin()
cities = ["Boston", "New York"]
print(f"\nThose living in {cities} cities:")
print(df[df["City"].isin(cities)])

# Filtering with string operations
print("\nNames starting with 'A':")
print(df[df["Name"].str.startswith("A")])

print("\nNames starting with 'J':")
print(df[df["Name"].str.startswith("J")])

# Range querying with Between
print("\nThose aged between 25-35:")
print(df[df["Age"].between(25, 35)])
```

# Data Grouping and Aggregation

```python
print("\n" + "=" * 50)
print("DATA GROUPING AND AGGREGATION")
print("=" * 50)

# Grouping operations
print("Average salary by city:")
print(df.groupby("City")["Salary"].mean())

print("\nDetailed statistics by city:")
print(
    df.groupby("City").agg(
        {"Salary": ["mean", "min", "max", "std"], "Age": ["mean", "count"]}
    )
)

# Multiple grouping
df["Gender"] = ["M", "M", "M", "F", "F"]
print("\nGrouping by City and Gender:")
print(df.groupby(["City", "Gender"])["Salary"].mean())
```

# Data Sorting

```python
print("\n" + "=" * 50)
print("DATA SORTING")
print("=" * 50)

# Sorting by single column
print("Ascending sort by Salary:")
print(df.sort_values("Salary"))

print("\nDescending sort by Salary:")
print(df.sort_values("Salary", ascending=False))

# Sorting by multiple columns
print("\nSort by City first, then Salary:")
print(df.sort_values(["City", "Salary"], ascending=[True, False]))
```

# Missing Data Operations

```python
print("\n" + "=" * 50)
print("MISSING DATA OPERATIONS")
print("=" * 50)

# Creating DataFrame with missing data
df_missing = df.copy()
df_missing.loc[0, "Age"] = None
df_missing.loc[2, "Salary"] = np.nan
df_missing.loc[4, "City"] = None

print("DataFrame with missing data:")
print(df_missing)

print("\nMissing data check:")
print(df_missing.isnull())

print("\nMissing data counts:")
print(df_missing.isnull().sum())

# Cleaning missing data
df_clean = df_missing.dropna()  # Drop rows with missing data
print("\nMissing data dropped:")
print(df_clean)

# Filling missing data
df_filled = df_missing.fillna(
    {
        "Age": df_missing["Age"].mean(),
        "Salary": df_missing["Salary"].median(),
        "City": "Unknown",
    }
)
print("\nMissing data filled:")
print(df_filled)
```

# DataFrame Merging

```python
print("\n" + "=" * 50)
print("DATAFRAME MERGING OPERATIONS")
print("=" * 50)

# Let's create two DataFrames
df1 = pd.DataFrame(
    {
        "ID": [1, 2, 3, 4],
        "Name": ["John", "Jill", "Jack", "Jenny"],
        "Department": ["IT", "HR", "IT", "Finance"],
    }
)

df2 = pd.DataFrame(
    {
        "ID": [1, 2, 3, 5],
        "Salary": [50000, 60000, 70000, 80000],
        "Experience": [3, 5, 8, 10],
    }
)

print("DF1:")
print(df1)
print("\nDF2:")
print(df2)

# Inner join
inner_merge = pd.merge(df1, df2, on="ID", how="inner")
print("\nInner Join:")
print(inner_merge)

# Left join
left_merge = pd.merge(df1, df2, on="ID", how="left")
print("\nLeft Join:")
print(left_merge)

# Concatenation
df3 = pd.DataFrame(
    {"ID": [5, 6], "Name": ["Jimmy", "Jony"], "Department": ["IT", "HR"]}
)

merged = pd.concat([df1, df3], ignore_index=True)
print("\nMerging with Concat:")
print(merged)
```

```python
# scenario 2 : data quality control
print("\n" + "=" * 50)
print("SCENARIO 2: DATA QUALITY CONTROL")
print("=" * 50)


def data_quality_control(dataframe):
    """Check data quality"""
    print("✅ DATA QUALITY REPORT")
    print("=" * 35)

    # Size info
    print(f"Row count: {len(dataframe)}")
    print(f"Column count: {len(dataframe.columns)}")

    # Missing data analysis
    missing_data = dataframe.isnull().sum()
    print("\nMISSING DATA ANALYSIS:")
    for column, missing in missing_data.items():
        if missing > 0:
            print(f"  ⚠️  {column}: {missing} missing data")
        else:
            print(f"  ✅ {column}: No missing data")

    # Data types
    print("\nDATA TYPES:")
    for column, tip in dataframe.dtypes.items():
        print(f"  {column}: {tip}")

    # Unique values
    print("\nUNIQUE VALUE COUNTS:")
    for column in dataframe.columns:
        unique = dataframe[column].nunique()
        print(f"  {column}: {unique} unique values")


data_quality_control(df)
```

# File Reading Example

```python
import pandas as pd

# Read CSV file
sales = pd.read_csv("sales_data.csv", encoding="utf-8")

# Inspect data
print("First 5 rows:")
print(sales.head())

print("\nData info:")
print(sales.info())

# Cleaning operations
sales["date"] = pd.to_datetime(sales["date"])
sales["month"] = sales["date"].dt.month
sales["year"] = sales["date"].dt.year

# Clean missing data
sales = sales.dropna()

# Monthly sales analysis
monthly_sales = (
    sales.groupby(["year", "month"])
    .agg({"amount": "sum", "customer_id": "nunique", "order_id": "count"})
    .reset_index()
)

monthly_sales.columns = [
    "Year",
    "Month",
    "Total Sales",
    "Unique Customer",
    "Total Order",
]

# Write to Excel
with pd.ExcelWriter("sales_analysis.xlsx") as writer:
    sales.to_excel(writer, sheet_name="Raw Data", index=False)
    monthly_sales.to_excel(writer, sheet_name="Monthly Summary", index=False)

print("Analysis completed and files saved!")
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random


# Random data creation functions
def random_date(start=datetime(2023, 1, 1), end=datetime(2024, 1, 1)):
    """Create random date"""
    diff = end - start
    random_day = random.randint(0, diff.days)
    return start + timedelta(days=random_day)


def random_product():
    """Create random product name"""
    products = [
        "Laptop",
        "Phone",
        "Tablet",
        "Headphone",
        "Mouse",
        "Keyboard",
        "Monitor",
        "Printer",
        "Camera",
        "Speaker",
        "Hard Disk",
        "SSD",
    ]
    return random.choice(products)


def random_city():
    """Create random city name"""
    cities = [
        "Istanbul",
        "Ankara",
        "Izmir",
        "Bursa",
        "Antalya",
        "Adana",
        "Konya",
        "Gaziantep",
        "Kayseri",
        "Mersin",
        "Diyarbakir",
        "Eskisehir",
    ]
    return random.choice(cities)


def random_category(product):
    """Determine category based on product"""
    category_map = {
        "Laptop": "Electronics",
        "Phone": "Electronics",
        "Tablet": "Electronics",
        "Headphone": "Accessory",
        "Mouse": "Accessory",
        "Keyboard": "Accessory",
        "Monitor": "Hardware",
        "Printer": "Hardware",
        "Camera": "Electronics",
        "Speaker": "Accessory",
        "Hard Disk": "Hardware",
        "SSD": "Hardware",
    }
    return category_map.get(product, "Other")


# Create dataset
np.random.seed(42)  # For reproducibility
random.seed(42)

sales = []

for i in range(1000):
    product = random_product()
    category = random_category(product)
    date = random_date()
    city = random_city()
    quantity = random.randint(1, 5)
    unit_price = round(random.uniform(50, 2000), 2)
    total_amount = round(quantity * unit_price, 2)

    # Add missing data in some rows (for realism)
    customer_age = random.randint(18, 70) if random.random() > 0.1 else np.nan
    customer_gender = (
        random.choice(["Male", "Female"]) if random.random() > 0.08 else np.nan
    )

    sales.append(
        {
            "sale_id": i + 1,
            "customer_id": f"M{random.randint(1000, 9999)}",
            "product_name": product,
            "category": category,
            "quantity": quantity,
            "unit_price": unit_price,
            "total_amount": total_amount,
            "date": date.strftime("%Y-%m-%d"),
            "city": city,
            "customer_age": customer_age,
            "customer_gender": customer_gender,
            "payment_type": random.choice(["Credit Card", "Cash", "Transfer"]),
            "discount_rate": round(random.uniform(0, 0.3), 2),  # 0-30% discount
        }
    )

# Create DataFrame
df = pd.DataFrame(sales)

# Save to CSV file
df.to_csv("sales_data.csv", index=False, encoding="utf-8")

print("CSV file created successfully!")
print(f"Created dataset size: {df.shape}")
print("\nFirst 5 rows:")
print(df.head())
print("\nData types:")
print(df.dtypes)
print("\nBasic statistics:")
print(df.describe())
```

# Using the Created Data

```python
# Reading the file
df_read = pd.read_csv("sales_data.csv", encoding="utf-8")

# Inspecting data
print("Dataset size:", df_read.shape)
print("\nFirst 10 rows:")
print(df_read.head(10))

print("\nColumns:")
print(df_read.columns.tolist())

print("\nMissing data:")
print(df_read.isnull().sum())

# Basic analyses
print("\nSales by City:")
print(df_read.groupby("city")["total_amount"].sum().sort_values(ascending=False))

print("\nSales by Category:")
print(df_read.groupby("category")["total_amount"].sum())

# Date conversion
df_read["date"] = pd.to_datetime(df_read["date"])
df_read["month"] = df_read["date"].dt.month
df_read["year"] = df_read["date"].dt.year

print("\nMonthly Sales:")
monthly_sales = df_read.groupby(["year", "month"])["total_amount"].sum()
print(monthly_sales)
```

# Creating Excel File with Same Data

```python
# Creating Excel file
with pd.ExcelWriter("sales_analysis.xlsx") as writer:
    df_read.to_excel(writer, sheet_name="All Data", index=False)

    # Summary tables
    city_summary = (
        df_read.groupby("city")
        .agg({"total_amount": "sum", "sale_id": "count", "unit_price": "mean"})
        .round(2)
    )
    city_summary.columns = ["Total Sales", "Sales Count", "Average Price"]
    city_summary.to_excel(writer, sheet_name="City Summary")

    category_summary = df_read.groupby("category").agg(
        {"total_amount": "sum", "sale_id": "count", "quantity": "sum"}
    )
    category_summary.columns = ["Total Sales", "Sales Count", "Total Quantity"]
    category_summary.to_excel(writer, sheet_name="Category Summary")

print("Excel files created!")
```
