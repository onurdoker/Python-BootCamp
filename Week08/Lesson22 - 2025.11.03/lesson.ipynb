# 🐼 Pandas Library:

Pandas is one of the most fundamental and powerful tools in Python for data science and analysis.

🌟 **What is Pandas?**

Pandas is an open-source library developed for data analysis and manipulation in the Python programming language. It basically provides fast and flexible data structures that make working with data very easy.

**Basic Data Structures:**
1. Series:

    A one-dimensional, labeled (indexed) data structure.

    Can be thought of as a single column.

2. DataFrame:

    A two-dimensional data structure with labeled (indexed) columns and rows.

    Similar to a table or Excel/SQL database structure.

    It is the most used data structure in Pandas.

📈 **Why Use Pandas?**

Pandas is indispensable for overcoming challenges encountered when working with real-world data.

*  Data Loading and Reading: It can easily load data from different formats (CSV, Excel, SQL, JSON, etc.) into a DataFrame structure.

* Data Cleaning:

    * Detecting, filling, or deleting missing values (NaN).

    * Converting data types (converting text to numbers).

    * Removing duplicate rows.

* Data Manipulation and Transformation:

    * Filtering, selecting, and slicing data.

    * Adding/removing columns, deleting rows.

    * Grouping data (groupby) and aggregating (mean, sum, count, etc.).

    * Merging, appending, or joining datasets (merge, join, concatenate).

* Data Analysis: Performing statistical summaries and analyses, calculating correlations.

🤝 **Relationship with NumPy and Other Libraries**

Pandas is at the center of the data science ecosystem in Python and works integrated with other libraries.

🐘 1 . **Relationship with NumPy (Foundation)**

* NumPy (Numerical Python) is the base Python library for high-performance numerical calculations.

* Speed and Performance: All high-performance mathematical and vector operations are provided by NumPy. While NumPy is the foundation of scientific computing; Pandas is a user-friendly high-level tool that enables tabular data analysis.

🖼️ 2 . Relationship with Matplotlib and Seaborn (Visualization)

* Role: After cleaning and preparing data with Pandas, you can switch to visualization with these libraries in a single line of code.

🤖 3 . Relationship with Scikit-learn and Deep Learning Libraries

* Role: Pandas performs the data preparation and preprocessing steps required for Machine Learning (ML).

* Flow: Data is cleaned and transformed with Pandas; when it is ready for modeling, it is easily converted into NumPy arrays and presented as input to Scikit-learn or deep learning libraries (TensorFlow, PyTorch).

# Pandas Series

A single-column labeled array. Holds both data and index information.
Examples with different types (int, float, string, date).

````python
# 1) Creating a simple Series (using a regular list)
s = pd.Series([10, 20, 30, 40])
# s is a Series object; each element has an index (0,1,2,...)
print("\n#1 - s:\n", s)
# Explanation: Output shows; index + values

# 2) Creating with assigned index
s2 = pd.Series([100, 200, 300], index=["a", "b", "c"])
print("\n#2 - b:\n", s2)
# Selecting by index labels:
print(s2["b"])  # 200

# 3) Creating Series from Dictionary (keys become index)
data = {"istanbul": 15, "ankara": 5, "izmir": 8}
pop = pd.Series(data)
print("\n#3 - pop\n", pop)
# If we query, a non-existent label returns NaN:
print(pop.get("antalya"))  # None or NaN; .get allows safe retrieval of None/NA

# 4) Mathematical operation with Series (broadcast)
s3 = pd.Series([1, 2, 3], index=["x", "y", "z"])
print(
    "\n#4 - Mathematical operations:\n", s3 * 10
)  # all values multiplied by 10 (index preserved)

# 5) Checking Series type and values
print(s3.dtype)  # dtype: int64 etc.
print(s3.index)  # Index(['x','y','z'], dtype='object')

# Series is actually a structure similar to a numpy array, but also holds index names.
# When selecting with index labels, Python dictionary logic and NumPy logic mix — pay attention to differences.
````

# Indexing Operations in Series

````python
# Creating Series
series = pd.Series([10, 20, 30, 40, 50], index=["a", "b", "c", "d", "e"])
print(series)
# a    10
# b    20
# c    30
# d    40
# e    50
# dtype: int64
````

# Basic Indexing in Series

````python
# Accessing a single value
print(series["b"])  # 20

# Accessing multiple values
print(series[["a", "c", "e"]])
# a    10
# c    30
# e    50
# dtype: int64

# Slice operation
print(series["b":"d"])
# b    20
# c    30
# d    40
# dtype: int64
````

# Indexing with .loc

````python
# Label-based indexing with .loc
# Single label
print(series.loc["c"])  # 30

# Multiple labels
print(series.loc[["a", "d"]])
# a    10
# d    40
# dtype: int64

# Slice operation (inclusive)
print(series.loc["b":"d"])
# b    20
# c    30
# d    40
# dtype: int64
````

# Indexing with .iloc

````python
# Position-based indexing with .iloc
# Single position
print(series.iloc[2])  # 30

# Multiple positions
print(series.iloc[[0, 3, 4]])
# a    10
# d    40
# e    50
# dtype: int64

# Slice operation
print(series.iloc[1:4])
# b    20
# c    30
# d    40
# dtype: int64
````

# Conditional Operations in Series

````python
# Filtering by condition
series = pd.Series([10, 25, 30, 45, 50], index=["a", "b", "c", "d", "e"])

# Values greater than 30
print(series[series > 30])
# d    45
# e    50
# dtype: int64

# Multiple conditions
print(series[(series > 20) & (series < 50)])
# b    25
# c    30
# d    45
# dtype: int64

# Equality to specific values
print(series.isin([10, 30, 50]))
print(series[series.isin([10, 30, 50])])
# a    10
# c    30
# e    50
# dtype: int64
````

# Creating Multi-Level Index

````python
# Creating multi-level index
arrays = [["A", "A", "B", "B"], [1, 2, 1, 2]]
index = pd.MultiIndex.from_arrays(arrays, names=["letter", "number"])
multi_series = pd.Series([10, 20, 30, 40], index=index)

print(index)
print(multi_series)
# letter  number
# A       1         10
#         2         20
# B       1         30
#         2         40
# dtype: int64

# Multi-level indexing
print(multi_series.loc["A"])
# number
# 1    10
# 2    20
# dtype: int64

print(multi_series.loc[("A", 2)])  # 20
print(multi_series.xs("B", level="letter"))
# number
# 1    30
# 2    40
# dtype: int64
````

# Conditional Indexing Example: Time-Based Examination

````python
# Let's create a large series
large_series = pd.Series(range(1000000))
print(f"Series size: {len(large_series)}")

# 1. SLOW Method: Chained indexing
start = time.time()
result1 = large_series[large_series > 500000][:10]
duration1 = time.time() - start
print(f"1. Chaining method duration: {duration1:.6f} seconds")

# 2. MEDIUM SPEED: Pandas indexing with .loc
start = time.time()
result2 = large_series.loc[large_series > 500000].iloc[:10]
duration2 = time.time() - start
print(f"2. Pandas .loc method duration: {duration2:.6f} seconds")

# 3. FASTEST: Working with pure numpy
start = time.time()
result3 = large_series.values[large_series.values > 500000][:10]
duration3 = time.time() - start
print(f"3. Pure numpy method duration: {duration3:.6f} seconds")

print("\nSpeed comparison:")
print(f"Numpy is {duration1 / duration3:.1f} times faster than chaining")
print(f"Numpy is {duration2 / duration3:.1f} times faster than pandas .loc")
````

````python
# Let's examine memory usage and data types
print("\nData structure analysis:")
print(f"Pandas Series type: {type(large_series)}")
print(f"Numpy array type: {type(large_series.values)}")
print(f"Pandas index size: {large_series.index.nbytes} bytes")
print(f"Pandas values size: {large_series.values.nbytes} bytes")

# Operation details
print("\nOperation complexity:")
print("1. Chaining: Pandas filter → Pandas slice → New Series")
print("2. Pandas .loc: Pandas optimization → New Series")
print("3. Numpy: Pure array operation → Array slice → Array")
````

````python
# Let's do a more realistic test
print("\n" + "=" * 50)
print("REALISTIC SCENARIO: More complex filter")
print("=" * 50)

# Series filled with random numbers
np.random.seed(42)
realistic_series = pd.Series(np.random.randint(0, 1000000, 1000000))

# 1. Chaining
start = time.time()
result1 = realistic_series[(realistic_series > 500000) & (realistic_series % 2 == 0)][
    :20
]
duration1 = time.time() - start

# 2. Pandas .loc
start = time.time()
result2 = realistic_series.loc[
    (realistic_series > 500000) & (realistic_series % 2 == 0)
].iloc[:20]
duration2 = time.time() - start

# 3. Numpy
start = time.time()
values = realistic_series.values
mask = (values > 500000) & (values % 2 == 0)
result3 = values[mask][:20]
duration3 = time.time() - start

print(f"Chaining: {duration1:.6f}s")
print(f"Pandas .loc: {duration2:.6f}s")
print(f"Numpy: {duration3:.6f}s")
print(f"Numpy is {duration2 / duration3:.1f} times faster than pandas .loc")
````

# Concatenating Series
Combining multiple Pandas Series into a single Series.
This operation is usually done vertically (one after another).

Think of it like:

Series 1: [10, 20] → indices [a, b]
Series 2: [30, 40] → indices [c, d]
When combined: [10, 20, 30, 40] → indices [a, b, c, d]

The `pd.concat()` function is used to combine series (and DataFrames) in Pandas.

`pd.concat([series1, series2, series3, ...])`

````python
series1 = pd.Series([1, 2], index=["a", "b"])
series2 = pd.Series([3, 4], index=["c", "d"])

combined = pd.concat([series1, series2])
print(combined)
````

**Combining Series with Same Index**
If two series have the same index values, concat adds them as separate rows (i.e., allows duplicate indices):

````python
series1 = pd.Series([10, 20], index=["x", "y"])
series2 = pd.Series([30, 40], index=["x", "y"])

result = pd.concat([series1, series2])
print(result)

# In the above case, indices are duplicated. If you don't want this, you can use reset_index(drop=True) after concatenation:

result = pd.concat([series1, series2], ignore_index=True)
# Output: 0:10, 1:20, 2:30, 3:40
print(result)
````

**Combining Multiple Series**

````python
s1 = pd.Series([1])
s2 = pd.Series([2])
s3 = pd.Series([3])
s4 = pd.Series([4])

all_series = pd.concat([s1, s2, s3, s4], ignore_index=True)
print(all_series)
````

# Combining Series Side-by-Side

Series are single-column, so combining them side-by-side with concat usually creates a DataFrame.

````python
series1 = pd.Series([1, 2], name="A")
series2 = pd.Series([3, 4], name="B")
series3 = pd.Series([5, 6], name="C")

df = pd.concat([series1, series2, series3], axis=1)
print(df)
# Note: Naming series is necessary for column names in side-by-side concatenation.

# axis=0 → vertical (default)
# axis=1 → horizontal (as columns)
````

# Arithmetic Operations in Series

Arithmetic operations in Pandas Series are applied to each element individually.

For example:

* series * 2 → Multiplies every number in the Series by 2.
* series1 + series2 → Adds elements at the same index.
These operations are automatically aligned: Pandas matches indices; puts NaN for non-matching places.

````python
prices = pd.Series([10, 20, 30], index=["apple", "pear", "banana"])
print("Original:\n", prices)

# Let's add 10% raise → i.e., multiply each price by 1.1
new_prices = prices * 1.1
print("\nPrices with raise:")
print(new_prices)
````

````python
# Let's create example series
series1 = pd.Series([10, 20, 30, 40, 50])
series2 = pd.Series([1, 2, 3, 4, 5])

print("Series1:", series1.values)
print("Series2:", series2.values)

# Addition
total = series1 + series2
print("Total:", total.values)

# Subtraction
difference = series1 - series2
print("Difference:", difference.values)

# Multiplication
product = series1 * series2
print("Product:", product.values)

# Division
division = series1 / series2
print("Division:", division.values)

# Power
power = series1**2
print("Square of Series1:", power.values)
````

````python
# Series with constant number operations Scalar operations
print("Series1 + 100:", (series1 + 100).values)
print("Series1 * 2:", (series1 * 2).values)
print("Series1 / 10:", (series1 / 10).values)
print("Series1 ** 0.5:", (series1**0.5).values)  # Square root
````

````python
# numpy math functions
print(series1)
print("Logarithm:", np.log(series1).values)
print("Exponential:", np.exp(series1 / 10).values)
print("Square root:", np.sqrt(series1).values)
print("Absolute value:", np.abs(pd.Series([-5, -2, 0, 3, 7])).values)
print("Sine:", np.sin(series1).values)
print("Rounding:", np.round(series1 / 3, 2).values)
````

# Operations with Different Indexed Series

````python
# Series with different indices
series_a = pd.Series([10, 20, 30], index=["a", "b", "c"])
series_b = pd.Series([5, 15, 25], index=["b", "c", "d"])

print("Series A:")
print(series_a)
print("\nSeries B:")
print(series_b)

# Index mismatch situation
print("\nSeries A + Series B:")
total_mismatch = series_a + series_b
print(total_mismatch)
print("NaN values occurred for non-common indices!")
````

````python
# Managing NaN values
print("\nAdding by setting NaN values to 0:")
total_fillna = series_a.add(series_b, fill_value=0)
print(total_fillna)

# Using fill_value parameter
print("\nUsing fill_value:")
product_fillna = series_a.mul(series_b, fill_value=1)
print(product_fillna)
````

````python
# Cumulative operations
series = pd.Series([2, 3, 1, 4, 5])
print("Original series:", series.values)
print("Cumulative sum:", series.cumsum().values)
print("Cumulative product:", series.cumprod().values)
print("Cumulative maximum:", series.cummax().values)
print("Cumulative minimum:", series.cummin().values)
````

````python
# Basic statistics
data_series = pd.Series([23, 45, 12, 67, 34, 89, 15, 42])

print("Series:", data_series.values)
print("Mean:", data_series.mean())
print("Median:", data_series.median())
print("Standard Deviation:", data_series.std())
print("Variance:", data_series.var())
print("Minimum:", data_series.min())
print("Maximum:", data_series.max())
print("Sum:", data_series.sum())
print("1st Quartile (Q1):", data_series.quantile(0.25))
print("3rd Quartile (Q3):", data_series.quantile(0.75))
````

````python
grades = pd.Series([45, 70, 85, 30], index=["John", "Jim", "Jack", "Jill"])

# Add 10 points to those below 50
grades[grades < 50] += 10
print(grades)
````

````python
# Comparison operations
series = pd.Series([15, 25, 35, 45, 55])

print("Series > 30:", (series > 30).values)
print("Series == 35:", (series == 35).values)
print("Series <= 40:", (series <= 40).values)

# Filtering with boolean indexing
print("Values greater than 30:", series[series > 30].values)
print("Values between 20 and 50:", series[(series > 20) & (series < 50)].values)
````

````python
# Applying custom function
def custom_function(x):
    return x**2 + 2 * x + 1


series = pd.Series([1, 2, 3, 4, 5])
result = series.apply(custom_function)
print("Custom function result:", result.values)

# Lambda functions
square = series.apply(lambda x: x**2)
print("Squaring with lambda:", square.values)
````

````python
print("\n" + "=" * 50)
print("REAL LIFE EXAMPLE: Sales Analysis")
print("=" * 50)

# Sales data
sales = pd.Series(
    [15000, 22000, 18000, 25000, 30000],
    index=["January", "February", "March", "April", "May"],
)
expenses = pd.Series(
    [8000, 9500, 12000, 11000, 14000],
    index=["January", "February", "March", "April", "May"],
)

print("Monthly Sales:")
print(sales)

print("\nMonthly Expenses:")
print(expenses)

# Profit calculation
profit = sales - expenses
print("\nMonthly Profit:")
print(profit)

# Percentage change
percentage_change = sales.pct_change() * 100
print("\nMonthly % Change in Sales:")
print(percentage_change.round(2))

# Cumulative sales
cumulative_sales = sales.cumsum()
print("\nCumulative Sales:")
print(cumulative_sales)

# Performance above average
above_average = sales > sales.mean()
print("\nMonths with Above Average Sales:")
print(sales[above_average])
````

````python
# Method chaining example
result = (
    pd.Series([10, 25, 35, 42, 58])
    .apply(lambda x: x * 1.1)  # 10% increase
    .round(1)  # Round
    .where(lambda x: x > 30)  # Take those greater than 30
    .dropna()  # Clean NaNs
)

print("Method chaining result:")
print(result)
````

````python
# Broadcast operations
series = pd.Series([1, 2, 3, 4, 5])

# Apply same operation to all elements
print("Series + 10:", (series + 10).values)
print("Series * 2:", (series * 2).values)

# Broadcast with other series
other_series = pd.Series([100])  # Single element series
print("Series + single element series:", (series + other_series).values)
````

* fill_value: Use fill_value parameter to prevent NaN values

* In-place Operation: Operations like series += 10 change the original series

# **Let's create a sample CSV file that we can use to test pandas operations**

````python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random


# Random data creation functions
def random_date(start=datetime(2023, 1, 1), end=datetime(2024, 1, 1)):
    """Create random date"""
    diff = end - start
    random_day = random.randint(0, diff.days)
    return start + timedelta(days=random_day)


def random_product():
    """Create random product name"""
    products = [
        "Laptop",
        "Phone",
        "Tablet",
        "Headphone",
        "Mouse",
        "Keyboard",
        "Monitor",
        "Printer",
        "Camera",
        "Speaker",
        "Hard Disk",
        "SSD",
    ]
    return random.choice(products)


def random_city():
    """Create random city name"""
    cities = [
        "Istanbul",
        "Ankara",
        "Izmir",
        "Bursa",
        "Antalya",
        "Adana",
        "Konya",
        "Gaziantep",
        "Kayseri",
        "Mersin",
        "Diyarbakir",
        "Eskisehir",
    ]
    return random.choice(cities)


def random_category(product):
    """Determine category based on product"""
    category_map = {
        "Laptop": "Electronics",
        "Phone": "Electronics",
        "Tablet": "Electronics",
        "Headphone": "Accessory",
        "Mouse": "Accessory",
        "Keyboard": "Accessory",
        "Monitor": "Hardware",
        "Printer": "Hardware",
        "Camera": "Electronics",
        "Speaker": "Accessory",
        "Hard Disk": "Hardware",
        "SSD": "Hardware",
    }
    return category_map.get(product, "Other")


# Create dataset
np.random.seed(42)  # For reproducibility
random.seed(42)

sales = []

for i in range(1000):
    product = random_product()
    category = random_category(product)
    date = random_date()
    city = random_city()
    quantity = random.randint(1, 5)
    unit_price = round(random.uniform(50, 2000), 2)
    total_amount = round(quantity * unit_price, 2)

    # Add missing data in some rows (for realism)
    customer_age = random.randint(18, 70) if random.random() > 0.1 else np.nan
    customer_gender = (
        random.choice(["Male", "Female"]) if random.random() > 0.08 else np.nan
    )

    sales.append(
        {
            "sale_id": i + 1,
            "customer_id": f"M{random.randint(1000, 9999)}",
            "product_name": product,
            "category": category,
            "quantity": quantity,
            "unit_price": unit_price,
            "total_amount": total_amount,
            "date": date.strftime("%Y-%m-%d"),
            "city": city,
            "customer_age": customer_age,
            "customer_gender": customer_gender,
            "payment_type": random.choice(["Credit Card", "Cash", "Transfer"]),
            "discount_rate": round(random.uniform(0, 0.3), 2),  # 0-30% discount
        }
    )

# Create DataFrame
df = pd.DataFrame(sales)

# Save to CSV file
df.to_csv("sales_data.csv", index=False, encoding="utf-8")

print("CSV file created successfully!")
print(f"Created dataset size: {df.shape}")
print("\nFirst 5 rows:")
print(df.head())
print("\nData types:")
print(df.dtypes)
print("\nBasic statistics:")
print(df.describe())
````

# Using the Created Data

````python
# Reading the file
df_read = pd.read_csv("sales_data.csv", encoding="utf-8")

# Inspecting data
print("Dataset size:", df_read.shape)
print("\nFirst 10 rows:")
print(df_read.head(10))

print("\nColumns:")
print(df_read.columns.tolist())

print("\nMissing data:")
print(df_read.isnull().sum())

# Basic analyses
print("\nSales by City:")
print(df_read.groupby("city")["total_amount"].sum().sort_values(ascending=False))

print("\nSales by Category:")
print(df_read.groupby("category")["total_amount"].sum())

# Date conversion
df_read["date"] = pd.to_datetime(df_read["date"])
df_read["month"] = df_read["date"].dt.month
df_read["year"] = df_read["date"].dt.year

print("\nMonthly Sales:")
monthly_sales = df_read.groupby(["year", "month"])["total_amount"].sum()
print(monthly_sales)
````

# Creating Excel File with Same Data

````python
# Creating Excel file
with pd.ExcelWriter("sales_analysis.xlsx") as writer:
    df_read.to_excel(writer, sheet_name="All Data", index=False)

    # Summary tables
    city_summary = (
        df_read.groupby("city")
        .agg({"total_amount": "sum", "sale_id": "count", "unit_price": "mean"})
        .round(2)
    )
    city_summary.columns = ["Total Sales", "Sales Count", "Average Price"]
    city_summary.to_excel(writer, sheet_name="City Summary")

    category_summary = df_read.groupby("category").agg(
        {"total_amount": "sum", "sale_id": "count", "quantity": "sum"}
    )
    category_summary.columns = ["Total Sales", "Sales Count", "Total Quantity"]
    category_summary.to_excel(writer, sheet_name="Category Summary")

print("Excel files created!")
````
